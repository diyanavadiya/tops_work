```python
# number guessing game 
import random

def play_game():
    secret = random.randint(1, 10)

    while True:
        guess = int(input("Enter a number (1-10): "))

        if guess == secret:
            print("🎉 Correct! You Won!")
            break
        elif guess < secret:
            print("Too Low!")
        else:
            print("Too High!")

play_game()
```
